```python
# Initialize the student database using a dictionary
# Key: Roll Number (Int) -> Value: ( (Name, Branch), [Marks] )
student_db = {
    101: (("Amit Sharma", "Computer Science"), [85, 90, 88]),
    102: (("Priya Patel", "Information Technology"), [78, 82, 80]),
    103: (("Rahul Verma", "Electronics"), [92, 88, 94])
}

def display_records(message):
    """Helper function to cleanly display all student records."""
    print(f"\n--- {message} ---")
    print(f"{'Roll No':<10} {'Name':<20} {'Branch':<25} {'Marks':<15}")
    print("-" * 70)
    for roll_no, (info_tuple, marks_list) in student_db.items():
        name, branch = info_tuple
        marks_str = ", ".join(map(str, marks_list))
        print(f"{roll_no:<10} {name:<20} {branch:<25} {marks_str:<15}")

# 1. Initial State
display_records("Initial Student Records")

# 2. Add a new student record
# Attributes: Roll Number = 104, Name = Ananya Das, Branch = Electrical, Marks = [89, 91, 85]
new_roll = 104
new_info = ("Ananya Das", "Electrical Engineering")  # Tuple
new_marks = [89, 91, 85]  # List
student_db[new_roll] = (new_info, new_marks)
print(f"\n[SUCCESS] Added student record for Roll No: {new_roll}")

# 3. Update the details of a student
# Updating Marks for Roll Number 102
update_roll = 102
if update_roll in student_db:
    # Lists are mutable, so we can modify the marks directly or reassign the full structure
    student_db[update_roll][1][0] = 95  # Changing the first subject mark to 95
    print(f"[SUCCESS] Updated marks for Roll No: {update_roll}")

# 4. Delete an existing student record
# Deleting Roll Number 103
delete_roll = 103
if delete_roll in student_db:
    del student_db[delete_roll]
    print(f"[SUCCESS] Deleted student record for Roll No: {delete_roll}")
else:
    print(f"[ERROR] Roll No {delete_roll} not found.")

# 5. Display the final student records
display_records("Final Student Records After Operations")
```
